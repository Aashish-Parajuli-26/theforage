```python
avg_sale=DA.groupby(['LIFESTAGE','PREMIUM_CUSTOMER']).agg(no_of_unit =('PROD_QTY','sum'),no_customers=('LYLTY_CARD_NBR','nunique') ).reset_index()
```

```python
avg_sale['AVG']=avg_sale['no_of_unit']/avg_sale['no_customers']
```

```python
avg_sale
```

```python
piv=avg_sale.pivot(index='LIFESTAGE', columns='PREMIUM_CUSTOMER', values='AVG')
fx,ax = plt.subplots(figsize=(15,10))
piv.plot(kind ='bar',ax=ax)
ax.set_xlabel('lifestage')
ax.set_ylabel('avg unit per customer')
ax.set_title('Avgera number of units per customers per lifestage')
plt.xticks(rotation = 0, ha='center')
plt.tight_layout()
plt.show()
```

```python
avg_price=DA.groupby(['LIFESTAGE','PREMIUM_CUSTOMER']).agg(total_sales=('TOT_SALES','sum'),total_units=('PROD_QTY','sum')).reset_index()
```

```python
avg_price['avg_price_perunit']=avg_price['total_sales']/avg_price['total_units']
```

```python
avg_price
```

```python
Pl=avg_price.pivot(index='LIFESTAGE',columns='PREMIUM_CUSTOMER',values='avg_price_perunit')
fig,ax=plt.subplots(figsize=(15,15))
Pl.plot(kind='bar',ax=ax)
ax.set_xlabel('Lifestage')
ax.set_ylabel('avg price per unit')
ax.set_title('Avg price per unit by lifestage and premium customer')
plt.legend(title='PREMIUM_CUSTOMER')
plt.xticks(rotation=0,ha='center')
plt.tight_layout()
plt.show()
```

```python
from scipy import stats
testing=DA[DA['LIFESTAGE'].isin(['MIDAGE SINGLES/COUPLES','YOUNG SINGLES/COUPLES'])]
```

```python
testing
```

```python
testing['PRICE_PER_UNIT']=testing['TOT_SALES']/testing['PROD_QTY']
testing
```

```python
mainstream = testing[testing['PREMIUM_CUSTOMER'] == 'Mainstream']['PRICE_PER_UNIT']
```

```python
mainstream
```

```python
budget_premium=testing[testing['PREMIUM_CUSTOMER'].isin(['Budget','Premium'])]['PRICE_PER_UNIT']
```

```python
t_stat,p_value=stats.ttest_ind(mainstream,budget_premium,equal_var=False)
print(f't-statistic:{t_stat},p-value:{p_value}')
```

```python
print(mainstream.mean(), budget_premium.mean())
```

```python
brand_segment=DA[(DA['LIFESTAGE']=='YOUNG SINGLES/COUPLES') & (DA['PREMIUM_CUSTOMER']=='Mainstream')]
```

```python
brand_segment
```

```python
brand_other_segment=DA[~((DA['LIFESTAGE']=='YOUNG SINGLES/COUPLES') & (DA['PREMIUM_CUSTOMER']=='Mainstream'))]
```

```python
brand_other_segment
```

```python
brand_segment_share=brand_segment['BRAND_NAME'].value_counts(normalize = True)
```

```python
brand_segment_share * 100
```

```python
brand_other_segment_share=brand_other_segment['BRAND_NAME'].value_counts(normalize = True)
```

```python
brand_other_segment_share * 100
```

```python
affinity=(brand_segment_share/brand_other_segment_share).sort_values(ascending=False)
affinity
```

```python
package_segment_share=brand_segment['WEIGHT'].value_counts(normalize = True)
package_other_segment_share=brand_other_segment['WEIGHT'].value_counts(normalize = True)
affinity_PACKAGE=(package_segment_share/package_other_segment_share).sort_values(ascending=False)
package_segment_share
package_other_segment_share
affinity_PACKAGE
```
